```r
baixar_texto <- function(url, arquivo) {

  pagina <- read_html(url)

  paragrafos <- html_elements(pagina, "p")

  texto <- html_text2(paragrafos)

  texto <- texto[trimws(texto) != ""]

  data.frame(
    arquivo = arquivo,
    paragrafo = 1:length(texto),
    texto = texto
  )
}
```

```r
docs <- rbind(
  baixar_texto(links[1], "Santos"),
  baixar_texto(links[2], "São Vicente"),
  baixar_texto(links[3], "Cubatão")
)
```

```r
limpar <- function(x) {

  x <- tolower(x)

  x <- iconv(
    x,
    from = "UTF-8",
    to = "ASCII//TRANSLIT"
  )

  x <- gsub("[^a-z0-9 ]", " ", x)

  x <- gsub("\\s+", " ", x)

  trimws(x)
}
```

```r
limpar(brutos[["p1"]])
```

```r
stopwords <- c(
  "de", "o", "a", "e", "um",
  "por", "como", "que", "da", "do"
)
```

```r
tok <- function(x) {

  unlist(
    strsplit(
      limpar(x),
      " "
    )
  )
}
```

```r
tok(brutos[["p1"]])
```

```r
sem_stop <- function(x) {

  t <- tok(x)

  t[!t %in% stopwords]
}
```

```r
sem_stop(brutos[["p1"]])
```

```r
install.packages("SnowballC")
```

```r
library(SnowballC)
```

```r
prep <- function(x) {

  termos <- sem_stop(x)

  wordStem(
    termos,
    language = "portuguese"
  )
}
```

```r
prep(brutos[["p1"]])
```

```r
postings <- list()

for (d in names(brutos)) {

  termos <- unique(
    prep(brutos[[d]])
  )

  for (termo in termos) {

    postings[[termo]] <- c(
      postings[[termo]],
      d
    )
  }
}
```

```r
length(postings)
```

```r
busca_AND <- function(consulta) {

  termos <- unique(
    prep(consulta)
  )

  listas <- lapply(
    termos,
    function(termo) postings[[termo]]
  )

  if (any(sapply(listas, is.null))) {
    return(character(0))
  }

  Reduce(
    intersect,
    listas
  )
}
```

```r
busca_AND("santos cidade")
```

```r
busca_OR <- function(consulta) {

  termos <- unique(
    prep(consulta)
  )

  listas <- lapply(
    termos,
    function(termo) postings[[termo]]
  )

  listas <- listas[
    !sapply(listas, is.null)
  ]

  if (length(listas) == 0) {
    return(character(0))
  }

  Reduce(
    union,
    listas
  )
}
```

```r
busca_OR("santos cidade")
```

```r
resultado_AND <- busca_AND("santos cidade")

resultado_OR <- busca_OR("santos cidade")

length(resultado_AND)

length(resultado_OR)
```

```r
resultado <- busca_AND("santos cidade")

docs[
  docs$id %in% resultado,
  c("id", "arquivo", "paragrafo", "texto")
]
```
