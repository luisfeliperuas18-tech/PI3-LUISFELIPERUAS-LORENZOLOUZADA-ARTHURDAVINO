```r
frases <- c(

  f1 = "o gato caca o rato",
  f2 = "o rato foge do gato",
  f3 = "o gato dorme"
)
tokens <- lapply(frases, function(x) {
  unlist(strsplit(tolower(x), "\\s+"))
})
todos <- unlist(tokens)
contagem <- sort(table(todos), decreasing = TRUE)
contagem
```
