# Corpus da Baixada Santista

Municípios escolhidos:
- Santos
- São Vicente
- Cubatão

```r
install.packages("rvest")

library(rvest)
```

```r
links <- c(
  "https://pt.wikipedia.org/wiki/Santos",
  "https://pt.wikipedia.org/wiki/S%C3%A3o_Vicente_(S%C3%A3o_Paulo)",
  "https://pt.wikipedia.org/wiki/Cubat%C3%A3o"
)

links
```

```r
baixar_texto <- function(url) {

  pagina <- read_html(url)

  paragrafos <- html_elements(pagina, "p")

  texto <- html_text2(paragrafos)

  paste(texto, collapse = " ")
}
```

```r
docs <- c(
  baixar_texto(links[1]),
  baixar_texto(links[2]),
  baixar_texto(links[3])
)
```

```r
length(docs)
```

```r
texto_completo <- paste(docs, collapse = " ")

tokens <- unlist(
  strsplit(
    tolower(texto_completo),
    "[^[:alpha:]À-ÿ]+"
  )
)

tokens <- tokens[tokens != ""]
```

```r
head(tokens, 20)
```

```r
vocabulario <- unique(tokens)

length(vocabulario)
```

```r
frequencia <- sort(
  table(tokens),
  decreasing = TRUE
)

head(frequencia, 10)
```

Análise dos termos mais frequentes:
Os 10 termos mais frequentes não são necessariamente informativos, pois aparecem principalmente artigos, preposições e conjunções, como "de", "a", "o" e "e". Essas palavras aparecem muitas vezes na língua portuguesa, independentemente do assunto abordado. Isso sugere que, na próxima etapa, seria interessante remover as chamadas stopwords para identificar palavras que representem melhor o conteúdo do corpus.
