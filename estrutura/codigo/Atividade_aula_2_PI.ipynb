RECRIANDO O CORPUS DA AULA 1:

```r
docs <- c(
  d1 = "recuperacao de informacao ordena documentos por relevancia",
  d2 = "o modelo de espaco vetorial representa documentos como vetores",
  d3 = "bm25 e um modelo probabilistico de ranqueamento de texto",
  d4 = "aprendizado estatistico fundamenta a recuperacao moderna",
  d5 = "o indice invertido acelera a busca em muitos documentos",
  d6 = "embeddings capturam a semantica de palavras e documentos",
  d7 = "a avaliacao mede a relevancia dos resultados da busca",
  d8 = "ciencia de dados combina estatistica e programacao"
)
```

TOKENIZANDO OS DOCUMENTOS:

```r
tokenizar <- function(texto) {
  texto <- tolower(texto)
  unlist(strsplit(texto, "\\s+"))
}

tokens <- lapply(docs, tokenizar)
```

MONTANDO O VOCABULÁRIO:

```r
vocab <- sort(unique(unlist(tokens)))
length(vocab)
```

CONSTRUINDO A MATRIZ TERMO DOCUMENTO:

```r
tdm <- sapply(tokens, function(tk) as.integer(table(factor(tk, levels = vocab))))
rownames(tdm) <- vocab
dim(tdm)
```

CALCULANDO O IDF:

```r
N <- ncol(tdm)
df <- rowSums(tdm > 0)
idf <- log(N / df)
```

CALCULANDO A MATRIZ TF-IDF:

```r
tfidf <- tdm * idf
round(tfidf[c("documentos", "modelo", "de"), ], 2)
```

IMPLEMENTANDO A FUNÇÃO COSSENO:

```r
cosseno <- function(a, b) {
  sum(a * b) / (sqrt(sum(a^2)) * sqrt(sum(b^2)))
}
```

CRIANDO A FUNÇÃO DE BUSCA:

```r
buscar <- function(consulta) {
  tokens_consulta <- tokenizar(consulta)
  vetor_consulta <- as.integer(table(factor(tokens_consulta, levels = vocab)))
  peso_consulta <- vetor_consulta * idf

  scores <- apply(tfidf, 2, function(vetor_doc) cosseno(peso_consulta, vetor_doc))
  sort(scores, decreasing = TRUE)
}
```

TESTANDO SIMILARIDADE:

```r
cat("Consulta 1: modelo de recuperacao\n")
print(round(buscar("modelo de recuperacao"), 3))

cat("\nConsulta 2: busca e relevancia dos documentos\n")
print(round(buscar("busca e relevancia dos documentos"), 3))

cat("\nConsulta 3: estatistica e ciencia de dados\n")
print(round(buscar("estatistica e ciencia de dados"), 3))
```
